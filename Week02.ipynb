```python
# Perbandingan skimage.io.imread() dan cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
import cv2
import matplotlib.pyplot as plt
from skimage import io

path_gambar = '/content/drive/MyDrive/PCVK2026/gambar1.jpg'

# Dibaca dengan skimage (otomatis RGB)
img_skimage = io.imread(path_gambar)

# Dibaca cv2 (BGR) lalu di-cvtColor ke RGB
img_cv2_bgr = cv2.imread(path_gambar)
img_cv2_rgb = cv2.cvtColor(img_cv2_bgr, cv2.COLOR_BGR2RGB)

# Dibaca cv2 (BGR) tanpa konversi (untuk pembuktian kanal tertukar)
img_cv2_raw = img_cv2_bgr.copy()

# Visualisasi
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].imshow(img_skimage)
axes[0].set_title("skimage.io.imread()\n(Sudah RGB)")

axes[1].imshow(img_cv2_rgb)
axes[1].set_title("cv2.cvtColor(BGR2RGB)\n(Sudah Konversi ke RGB)")
plt.show()

# Nilai Piksel pada koordinat (y=50, x=50)
print("Nilai Piksel Koordinat (50, 50)")
print("skimage (RGB)             :", img_skimage[50, 50])
print("cv2 dikonversi (RGB)      :", img_cv2_rgb[50, 50])
print("cv2 mentah tanpa cvt (BGR):", img_cv2_raw[50, 50])
```

```python
# Pertanyaan analisa: Apakah proses mengecilkan citra (downsampling) kemudian membesarkannya kembali ke ukuran semula (upscaling) akan mengubah nilai rata-rata intensitas piksel dan menghilangkan detail citra?
# Read gambar format grayscale
img_gray = cv2.imread('/content/drive/MyDrive/PCVK2026/self.jpeg', cv2.IMREAD_GRAYSCALE)
h_orig, w_orig = img_gray.shape

# statistik awal
mean_awal = np.mean(img_gray)
std_awal = np.std(img_gray)

# Downsampling (Mengecilkan ke 25% ukuran asli)
w_small, h_small = int(w_orig * 0.25), int(h_orig * 0.25)
img_small = cv2.resize(img_gray, (w_small, h_small), interpolation=cv2.INTER_AREA)

# Upsampling (Membesarkan kembali ke 100% ukuran asli)
img_restored = cv2.resize(img_small, (w_orig, h_orig), interpolation=cv2.INTER_CUBIC)

# Ukur statistik akhir
mean_akhir = np.mean(img_restored)
std_akhir = np.std(img_restored)
mae = np.mean(np.abs(img_gray.astype(float) - img_restored.astype(float)))

# Visualisasi Perbandingan
fig, axes = plt.subplots(1, 3, figsize=(12, 4))
axes[0].imshow(img_gray, cmap='gray')
axes[0].set_title(f"Asli ({w_orig}x{h_orig})\nMean: {mean_awal:.2f}")

axes[1].imshow(img_small, cmap='gray')
axes[1].set_title(f"Dikecilkan ({w_small}x{h_small})")

axes[2].imshow(img_restored, cmap='gray')
axes[2].set_title(f"Dikembalikan ({w_orig}x{h_orig})\nMean: {mean_akhir:.2f}")
plt.show()

print("=== HASIL EKSPERIMEN ===")
print(f"Rata-rata Intensitas Awal  : {mean_awal:.4f}")
print(f"Rata-rata Intensitas Akhir : {mean_akhir:.4f}")
print(f"Selisih Rata-Rata (Mean)   : {abs(mean_awal - mean_akhir):.4f}")
print(f"Mean Absolute Error (MAE)  : {mae:.4f} piksel")
print(f"Standar Deviasi Awal       : {std_awal:.4f}")
print(f"Standar Deviasi Akhir      : {std_akhir:.4f}")
```
